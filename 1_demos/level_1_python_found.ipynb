## **1 Python Dictionaries (Dicts)**
A **dictionary** stores data as key-value pairs, like this: "**key_1**": "**value_1**". Each value is accessed up by its key (a label), not by its position. It lives in-memory,while the program is running, and disapears when it stops. we can add and you can add, change, or remove entries at any time.


Dictionaries and JSON look alike but live in different worlds: a dict exists only in Python's memory, while JSON is text — the format APIs, files, and other languages actually exchange. This file covers the conversion between the two, using Python's json module: loads/dumps for strings, load/dump for files. Mastering this conversion is the first step toward sending and receiving data from any LLM API.

````python
dic_words = {
    "wordId": "5",
    "type": "Noun Phrase",
    "word": "Communal road"
}
````

## **3 JSON Module**

#### **1-2-1  json.loads(): Loading a JSON-formatted String**
**`json.loads()`** takes a string: just a sequence of text characters, nothing more. Then, it converts it into native **Python objects** (**dicts**, **lists**, **strings**, **numbers**, etc.). The **"s"** at the end stands for **string**, which is the only thing this function accepts as **input**. **``json.loads(f.read())``** is the **general tool** for any string. Here is its syntax when dealing with a json file on disk: 

```python
import json

with open ("file.json") as f:
    python_object = json.loads(f.read())

```
And **where that string** comes from doesn't matter. It can originate from: 
- A **json file**, just as below with **``f.read()``** -An **HTTP response body** from an API call
- A **hardcoded string** that is typed directly inside the code
- A **raw bytes** received over a network socket
- A **database column** pulled from a database

````python
import json

#
with open ("data.json") as f:
    words_data = json.loads(f.read())

#
print(type(words_data), words_data, sep=" *** ")
````

#### **1-2-2 Converting a Python Object into a JSON-Formatted String: json.dumps(obj)**
``json.dumps(obj)`` converts a **Python object** (**dict**, **list**, **str**, **int**, **float**, etc.) into a **JSON-formatted string** (**object**, **arraye**, **string**, **number**, etc.)the "s" stands for string). When working with agents, it is not frequent that

````python
import json

s=json.dumps(dic_words) # Conversion
print(s) # New string format
print(type(s)) # s is a string...
print(type(dic_words)) # wile dic_words remained a dict
````

### **2-1 Read JSON from a file - json.load(file)**

#### **1-2-2 Converting a JSON-Formatted String into a Python Object: json.loads(str)**
json.loads(s) converts a JSON-formatted string (the "s" stands for string) into a Python object. The JSON string is parsed, and the corresponding Python dictionary or list is returned.

````python
import json

with open ("data.json") as f:
    words_data = json.loads(f.read())

print(type(words_data), words_data, sep=" *** ")
````

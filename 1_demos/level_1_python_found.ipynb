## **1 Python Dictionaries (Dicts)**
A **dictionary** stores data as key-value pairs, like this: "**key_1**": "**value_1**". Each **value** (the **data**) is accessed via its **key** (which is the **identifier**). Dictionaries live **in-memory**, when the program is running, and disapear when it stops. we can add, remove, or update entries at any time.

**Dictionaries** and **JSON** look alike but live in different spaces: dicts exist only in **Python's memory**, while **JSON** is a text format that **APIs**, **files** and **other languages** use to exchange data. Here is how a dict looks like in Python:

```python
dic_name = {
    "key_1": "value_1",
    "key_2": "value_2",
    "key_3": "value_3"
}
```
Its equivalent exist in JSON, it's called : **"object"**. This is how it's structured: 

```json
json_object = {
    "key_1": "value_1",
    "key_2": "value_2",
    "key_3": "value_3"
}
```

However, let's explore a bit more Python dicts before reverting back to JSON. 

````python
# This is a hand-made dict in Pyhton (also called dict literal)
dict_words = {
    "type": "Noun phrase",
    "word": "Communal road",
    "example": "He is walking on the main communal road"
}
# 1-Printing the initial dict
print(f"A-The initial Dict: {dict_words}\n")

# 2-Showing all the keys from dict_words 
all_keys = dict_words.keys()
print(f"B-All the Keys of initial Dict: {all_keys}\n")

# 3-Adding a new pair to dic_words ("wordId": 1)
dict_words.update(wordId=1)
print(f"C-The Dict with a new Pair: {dict_words}\n")

# 4-Removing a pair from dict_words ("type": "Noun phrase")
del dict_words["type"]
print(f"D-The Dict without a Pair: {dict_words}\n")

# 5-Replacing a value with another one ("example": "The communal road is blocked")
dict_words["example"] = "The communal road is blocked"
print(f"E-The Dict gets a new Example: {dict_words}\n")
````

## **3 JSON Module**

#### **1-2-1  json.loads(): Loading a JSON-formatted String**
**`json.loads()`** takes a string: just a sequence of text characters, nothing more. Then, it converts it into native **Python objects** (**dicts**, **lists**, **strings**, **numbers**, etc.). The **"s"** at the end stands for **string**, which is the only thing this function accepts as **input**. **``json.loads(f.read())``** is the **general tool** for any string. Here is its syntax when dealing with a json file on disk: 

```python
import json

with open ("file.json") as f:
    python_object = json.loads(f.read())

```
And **where that string** comes from doesn't matter. It can originate from: 
- A **json file**, just as below with **``f.read()``** -An **HTTP response body** from an API call
- A **hardcoded string** that is typed directly inside the code
- A **raw bytes** received over a network socket
- A **database column** pulled from a database

````python
import json

#
with open ("data.json") as f:
    words_data = json.loads(f.read())

#
print(type(words_data), words_data, sep=" *** ")
````

#### **1-2-2 Converting a JSON-Formatted String into a Python Object: json.loads(str)**
json.loads(s) converts a JSON-formatted string (the "s" stands for string) into a Python object. The JSON string is parsed, and the corresponding Python dictionary or list is returned.

````python
import json

with open ("data.json") as f:
    words_data = json.loads(f.read())

print(type(words_data), words_data, sep=" *** ")
````
